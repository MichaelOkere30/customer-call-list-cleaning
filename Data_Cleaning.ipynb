```python
import pandas as pd
```

1. Load the Data

```python
df = pd.read_excel('Customer Call List.xlsx')
```

```python
df
```

```python
df.info()
```

2. Remove Duplicates

```python
df = df.drop_duplicates()
df
```

3. Drop Unnecessary Columns   
Dropping 'Not_Useful_Column' as it adds no value to the analysis

```python
df = df.drop(columns='Not_Useful_Column')
df
```

4. Clean String Columns (Last Name)   
Strip unwanted punctuation from the Last_Name column

```python
df['Last_Name'] = df['Last_Name'].str.strip("123._/")
df
```

5. Standardize Yes/No Columns

```python
yes_no_mapping = {'Yes': 'Y', 'No': 'N'}
df['Paying Customer'] = df['Paying Customer'].replace(yes_no_mapping)
df['Do_Not_Contact'] = df['Do_Not_Contact'].replace(yes_no_mapping)
df
```

6. Clean and Format Phone Numbers
Remove all non-alphanumeric characters

```python
df['Phone_Number'] = df['Phone_Number'].replace('[^a-zA-Z0-9]', '', regex=True)
# Convert to string to allow string slicing 
df['Phone_Number'] = df['Phone_Number'].apply(lambda x: str(x))
# Format as XXX-XXX-XXX
df['Phone_Number'] = df['Phone_Number'].apply(lambda x: x[0:3] + '-' + x[3:6] + '-' + x[6:9])
df
```

7. Parse Address Column

```python
# Split the address column into Street, State, and Zip Code
df[['Street_Address', 'State', 'Postal_Code']] = df['Address'].str.split(',', n=2, expand=True)
df
```

8. Handle Missing/Null Values

```python
df['Phone_Number'] = df['Phone_Number'].str.replace('nan--', '')
df['Phone_Number'] = df['Phone_Number'].str.replace('Na--', '')
df = df.fillna('')

df['Address'] = df['Address'].str.replace('N/a', '')
df['Paying Customer'] = df['Paying Customer'].str.replace('N/a', '')
df['Street_Address'] = df['Street_Address'].str.replace('N/a', '')

df
```

10. Drop row where "Do_Not_Contact" is Y

```python
for x in df.index:
    if df.loc[x, 'Do_Not_Contact']=='Y':
        df.drop(x, inplace=True)
df
```

11. Drop row "Phone_Number" is empty

```python
for x in df.index:
    if df.loc[x, 'Phone_Number']=='':
        df.drop(x, inplace=True)
df
```

12. Replace Empty Do_Not_Contact with 'N'

```python
df['Do_Not_Contact'] = df['Do_Not_Contact'].replace('', 'N')
df
```

13. Reset Index

```python
df.reset_index(drop=True)
```

14. Export Cleaned Data

```python
df.to_csv("Cleaned_Customer_Call_List.csv", index=False)
print("Data cleaning completed successfully.")
```
